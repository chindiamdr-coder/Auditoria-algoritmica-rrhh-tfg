```python
# AUDITORÍA ALGORÍTIMICA DE RRHH: PROTOCOLO DE MITIGACIÓN DE SESGOS

# 1. IMPORTACIÓN DE LIBRERÍAS PARA CINCIA DE DATOS

import pandas as pd    # Pandas para la manipulación de datos en tablas
import numpy as np     # Numpy para la generación de datos sintéticos y cálculos varios
import matplotlib.pyplot as plt     # Matplotlib para la creación de gráficos para la auditoría

# Configuración de la visualización
plt.style.use('default')
plt.rcParams['figure.figsize'] = (8, 4)

print("Librerías cargadas correctamente")
```

```python
# 2. VISUALIZACIÓN DE RESULTADOS: IMPACTO DISPAR (PRE-MITIGACIÓN)

# Simulación de los datos obtenidos en la auditoría de CV's gemelos
categorias = ['Mujer (grupo test)', 'hombre (grupo control)']
tasas_seleccion = [32, 92]    # El sesgo detectado

#Creación de la figura y del eje ax
fig, ax = plt.subplots(figsize=(9, 5))
ax.set_facecolor('white')
barras = ax.bar(categorias, tasas_seleccion, color=['#ff9999', '#66b3ff'], edgecolor='black')

#Línea roja para el 80% (Regla de los 4/5)
limite_80 = 92 * 0.8
plt.axhline(y=limite_80, color='red', linestyle='--', linewidth=2, label='Límite de impacto dispar (80%)')

# Títulos y etiquetas
ax.set_title('Gráfico 1. Tasa de selección por género en puestos técnicos', fontsize=14, pad=20)
ax.set_ylabel('Candidatos seleccionados (%)', fontsize=12)
ax.set_ylim(0, 100)
ax.legend()

# Etiquetas de porcentaje sobre las barras
for barra in barras:
    yval = barra.get_height()
    ax.text(barra.get_x() + barra.get_width()/2, yval + 2, f'{yval}%', ha='center', va='bottom', fontweight='bold')

plt.show()
```
